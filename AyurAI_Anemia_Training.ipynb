```python
import os
import shutil
from sklearn.model_selection import train_test_split

# Source folders
anemic_dir = '/content/CP-AnemiC/Anemic'
non_anemic_dir = '/content/CP-AnemiC/Non-anemic'

# Collect image paths
anemic_files = [
    os.path.join(anemic_dir, f)
    for f in os.listdir(anemic_dir)
    if f.lower().endswith(('.png', '.jpg', '.jpeg'))
]

non_anemic_files = [
    os.path.join(non_anemic_dir, f)
    for f in os.listdir(non_anemic_dir)
    if f.lower().endswith(('.png', '.jpg', '.jpeg'))
]

all_files = anemic_files + non_anemic_files
all_labels = [1] * len(anemic_files) + [0] * len(non_anemic_files)

# First split: 70% train, 30% temporary
train_files, temp_files, train_labels, temp_labels = train_test_split(
    all_files,
    all_labels,
    test_size=0.30,
    stratify=all_labels,
    random_state=42
)

# Second split: 15% validation, 15% test
val_files, test_files, val_labels, test_labels = train_test_split(
    temp_files,
    temp_labels,
    test_size=0.50,
    stratify=temp_labels,
    random_state=42
)

print("Train:", len(train_files))
print("Validation:", len(val_files))
print("Test:", len(test_files))
```

```python
import os
import shutil

split_root = '/content/CP-AnemiC_split'

# Create folders
for split in ['train', 'val', 'test']:
    for class_name in ['Anemic', 'Non-anemic']:
        os.makedirs(
            os.path.join(split_root, split, class_name),
            exist_ok=True
        )

# Function to copy files
def copy_files(files, labels, split):
    for file_path, label in zip(files, labels):
        class_name = 'Anemic' if label == 1 else 'Non-anemic'
        destination = os.path.join(
            split_root,
            split,
            class_name,
            os.path.basename(file_path)
        )
        shutil.copy2(file_path, destination)

# Copy each split
copy_files(train_files, train_labels, 'train')
copy_files(val_files, val_labels, 'val')
copy_files(test_files, test_labels, 'test')

print("Dataset split folders created successfully!")
```

```python
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

# ImageNet normalization
mean = [0.485, 0.456, 0.406]
std = [0.229, 0.224, 0.225]

# Training transformations
train_transform = transforms.Compose([
    transforms.Lambda(lambda img: img.convert("RGB")),
    transforms.RandomResizedCrop(224, scale=(0.85, 1.0)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(5),
    transforms.ColorJitter(
        brightness=0.10,
        contrast=0.10,
        saturation=0.05,
        hue=0.02
    ),
    transforms.ToTensor(),
    transforms.Normalize(mean, std)
])

# Validation/Test transformations
eval_transform = transforms.Compose([
    transforms.Lambda(lambda img: img.convert("RGB")),
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(mean, std)
])

# Dataset folders
train_dataset = datasets.ImageFolder(
    '/content/CP-AnemiC_split/train',
    transform=train_transform
)

val_dataset = datasets.ImageFolder(
    '/content/CP-AnemiC_split/val',
    transform=eval_transform
)

test_dataset = datasets.ImageFolder(
    '/content/CP-AnemiC_split/test',
    transform=eval_transform
)

print("Classes:", train_dataset.classes)
print("Class mapping:", train_dataset.class_to_idx)

print("Train:", len(train_dataset))
print("Validation:", len(val_dataset))
print("Test:", len(test_dataset))
```

```python
import torch
from torch.utils.data import DataLoader

BATCH_SIZE = 32

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

print("Train batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Test batches:", len(test_loader))
```

```python
import torch
import torch.nn as nn
from torchvision.models import efficientnet_b0, EfficientNet_B0_Weights

# Use ImageNet-pretrained EfficientNet-B0
weights = EfficientNet_B0_Weights.DEFAULT
model = efficientnet_b0(weights=weights)

# Replace final classifier
num_features = model.classifier[1].in_features
model.classifier[1] = nn.Linear(num_features, 2)

# Move model to GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)

print("Device:", device)
print("Final classifier:", model.classifier[1])
```

```python
import torch

# ImageFolder mapping:
# 0 = Anemic
# 1 = Non-anemic

class_counts = torch.tensor([297, 200], dtype=torch.float32)

class_weights = class_counts.sum() / (2.0 * class_counts)

class_weights = class_weights.to(device)

print("Class weights:", class_weights)
print("Anemic weight:", class_weights[0].item())
print("Non-anemic weight:", class_weights[1].item())
```

```python
import torch.nn as nn
import torch.optim as optim

# Weighted loss
criterion = nn.CrossEntropyLoss(weight=class_weights)

# Optimizer
optimizer = optim.AdamW(
    model.parameters(),
    lr=1e-4,
    weight_decay=1e-4
)

# Reduce learning rate when validation AUROC stops improving
scheduler = optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode='max',
    factor=0.5,
    patience=2
)

print("Loss:", criterion)
print("Optimizer:", optimizer)
print("Learning rate:", optimizer.param_groups[0]['lr'])
print("Weight decay:", optimizer.param_groups[0]['weight_decay'])
print("Scheduler:", scheduler)
```

```python
import torch
import os

checkpoint_path = os.path.join(
    checkpoint_dir,
    'best_model_checkpoint.pth'
)

def save_checkpoint(epoch, model, optimizer, scheduler, best_val_auc):
    checkpoint = {
        'epoch': epoch,
        'model_state_dict': model.state_dict(),
        'optimizer_state_dict': optimizer.state_dict(),
        'scheduler_state_dict': scheduler.state_dict(),
        'best_val_auc': best_val_auc
    }

    torch.save(checkpoint, checkpoint_path)

    print(f"Checkpoint saved: {checkpoint_path}")
```

```python
import numpy as np
from sklearn.metrics import roc_auc_score, accuracy_score, confusion_matrix

def calculate_metrics(labels, probabilities, predictions):
    labels = np.array(labels)
    probabilities = np.array(probabilities)
    predictions = np.array(predictions)

    accuracy = accuracy_score(labels, predictions)

    # ROC-AUC
    auc = roc_auc_score(labels, probabilities)

    # Confusion matrix:
    # [[TN, FP],
    #  [FN, TP]]
    cm = confusion_matrix(labels, predictions, labels=[0, 1])

    tn, fp, fn, tp = cm.ravel()

    sensitivity = tp / (tp + fn) if (tp + fn) > 0 else 0
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0

    return {
        'accuracy': accuracy,
        'auc': auc,
        'sensitivity': sensitivity,
        'specificity': specificity
    }

print("Metric function ready.")
```

```python
import time
import copy
import torch

NUM_EPOCHS = 15
EARLY_STOPPING_PATIENCE = 4

best_val_auc = 0.0
epochs_without_improvement = 0

history = {
    'train_loss': [],
    'train_accuracy': [],
    'val_loss': [],
    'val_accuracy': [],
    'val_auc': [],
    'val_sensitivity': [],
    'val_specificity': []
}

for epoch in range(NUM_EPOCHS):

    start_time = time.time()

    # =========================================================
    # TRAINING
    # =========================================================
    model.train()

    running_loss = 0.0
    train_correct = 0
    train_total = 0

    for images, labels in train_loader:

        images = images.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        optimizer.zero_grad()

        outputs = model(images)

        loss = criterion(outputs, labels)

        loss.backward()

        optimizer.step()

        running_loss += loss.item() * images.size(0)

        predictions = torch.argmax(outputs, dim=1)

        train_correct += (predictions == labels).sum().item()
        train_total += labels.size(0)

    train_loss = running_loss / train_total
    train_accuracy = train_correct / train_total

    # =========================================================
    # VALIDATION
    # =========================================================
    model.eval()

    val_running_loss = 0.0
    val_labels = []
    val_probabilities = []
    val_predictions = []

    with torch.no_grad():

        for images, labels in val_loader:

            images = images.to(device, non_blocking=True)
            labels = labels.to(device, non_blocking=True)

            outputs = model(images)

            loss = criterion(outputs, labels)

            val_running_loss += loss.item() * images.size(0)

            probabilities = torch.softmax(outputs, dim=1)

            predictions = torch.argmax(outputs, dim=1)

            val_labels.extend(labels.cpu().numpy())

            # Probability of class 1 = Non-anemic
            val_probabilities.extend(
                probabilities[:, 1].cpu().numpy()
            )

            val_predictions.extend(
                predictions.cpu().numpy()
            )

    val_loss = val_running_loss / len(val_dataset)

    val_metrics = calculate_metrics(
        val_labels,
        val_probabilities,
        val_predictions
    )

    val_accuracy = val_metrics['accuracy']
    val_auc = val_metrics['auc']
    val_sensitivity = val_metrics['sensitivity']
    val_specificity = val_metrics['specificity']

    # =========================================================
    # SAVE HISTORY
    # =========================================================
    history['train_loss'].append(train_loss)
    history['train_accuracy'].append(train_accuracy)

    history['val_loss'].append(val_loss)
    history['val_accuracy'].append(val_accuracy)
    history['val_auc'].append(val_auc)

    history['val_sensitivity'].append(val_sensitivity)
    history['val_specificity'].append(val_specificity)

    # =========================================================
    # LEARNING RATE SCHEDULER
    # =========================================================
    scheduler.step(val_auc)

    current_lr = optimizer.param_groups[0]['lr']

    # =========================================================
    # SAVE BEST MODEL
    # =========================================================
    if val_auc > best_val_auc:

        best_val_auc = val_auc
        epochs_without_improvement = 0

        save_checkpoint(
            epoch,
            model,
            optimizer,
            scheduler,
            best_val_auc
        )

        best_marker = " ⭐ BEST"

    else:

        epochs_without_improvement += 1
        best_marker = ""

    # =========================================================
    # PRINT RESULTS
    # =========================================================
    elapsed = time.time() - start_time

    print(
        f"\nEpoch [{epoch + 1}/{NUM_EPOCHS}]"
        f" | Time: {elapsed:.1f}s"
        f" | LR: {current_lr:.2e}"
    )

    print(
        f"Train Loss: {train_loss:.4f}"
        f" | Train Acc: {train_accuracy:.4f}"
    )

    print(
        f"Val Loss: {val_loss:.4f}"
        f" | Val Acc: {val_accuracy:.4f}"
        f" | Val AUC: {val_auc:.4f}"
    )

    print(
        f"Val Sensitivity: {val_sensitivity:.4f}"
        f" | Val Specificity: {val_specificity:.4f}"
        f"{best_marker}"
    )

    # =========================================================
    # EARLY STOPPING
    # =========================================================
    if epochs_without_improvement >= EARLY_STOPPING_PATIENCE:

        print(
            f"\nEarly stopping triggered after "
            f"{epoch + 1} epochs."
        )

        break

print("\nTraining finished.")
print(f"Best validation AUC: {best_val_auc:.4f}")
print(f"Best checkpoint: {checkpoint_path}")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)
import numpy as np
import torch

model.eval()

test_labels = []
test_probabilities_non_anemic = []
test_predictions = []

with torch.no_grad():

    for images, labels in test_loader:

        images = images.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        outputs = model(images)

        probabilities = torch.softmax(outputs, dim=1)

        predictions = torch.argmax(outputs, dim=1)

        test_labels.extend(labels.cpu().numpy())

        # Probability of class 1 = Non-anemic
        test_probabilities_non_anemic.extend(
            probabilities[:, 1].cpu().numpy()
        )

        test_predictions.extend(
            predictions.cpu().numpy()
        )

test_labels = np.array(test_labels)
test_predictions = np.array(test_predictions)
test_probabilities_non_anemic = np.array(
    test_probabilities_non_anemic
)

# ---------------------------------------------------------
# Convert to ANEMIA probability
# Class 0 = Anemic
# Class 1 = Non-anemic
# ---------------------------------------------------------

test_probabilities_anemic = 1.0 - test_probabilities_non_anemic

# For binary anemia screening:
# 1 = Anemic
# 0 = Non-anemic

test_anemia_labels = (test_labels == 0).astype(int)
test_anemia_predictions = (test_predictions == 0).astype(int)

# ---------------------------------------------------------
# Metrics
# ---------------------------------------------------------

accuracy = accuracy_score(
    test_anemia_labels,
    test_anemia_predictions
)

sensitivity = recall_score(
    test_anemia_labels,
    test_anemia_predictions,
    pos_label=1
)

specificity = recall_score(
    test_anemia_labels,
    test_anemia_predictions,
    pos_label=0
)

precision = precision_score(
    test_anemia_labels,
    test_anemia_predictions,
    pos_label=1,
    zero_division=0
)

f1 = f1_score(
    test_anemia_labels,
    test_anemia_predictions,
    pos_label=1,
    zero_division=0
)

roc_auc = roc_auc_score(
    test_anemia_labels,
    test_probabilities_anemic
)

cm = confusion_matrix(
    test_anemia_labels,
    test_anemia_predictions,
    labels=[0, 1]
)

print("=" * 50)
print("FINAL TEST SET RESULTS")
print("=" * 50)

print(f"Accuracy:    {accuracy:.4f}")
print(f"Sensitivity: {sensitivity:.4f}")
print(f"Specificity: {specificity:.4f}")
print(f"Precision:   {precision:.4f}")
print(f"F1 Score:    {f1:.4f}")
print(f"ROC-AUC:     {roc_auc:.4f}")

print("\nConfusion Matrix")
print("[[TN, FP],")
print(" [FN, TP]]")
print(cm)

print("\nClassification Report")
print(
    classification_report(
        test_anemia_labels,
        test_anemia_predictions,
        target_names=["Non-anemic", "Anemic"],
        zero_division=0
    )
)
```

```python
# ============================================================
# RESIZE EXPERIMENT — COMPLETE SELF-CONTAINED CELL
# ============================================================

import os
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim

from torchvision import transforms, datasets
from torch.utils.data import DataLoader
from sklearn.metrics import roc_auc_score

# -----------------------------
# DEVICE
# -----------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# -----------------------------
# PATHS
# -----------------------------
split_dir = "/content/CP-AnemiC_split"

checkpoint_path = (
    "/content/drive/MyDrive/Epics/AyurAI/datasets/"
    "SwasthyaAI_checkpoints/best_model_checkpoint.pth"
)

resize_checkpoint_path = (
    "/content/drive/MyDrive/Epics/AyurAI/datasets/"
    "SwasthyaAI_checkpoints/best_resize_model.pth"
)

# -----------------------------
# RESIZE TRAINING TRANSFORM
# -----------------------------
train_transform_resize = transforms.Compose([
    transforms.Lambda(lambda img: img.convert("RGB")),
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(5),
    transforms.ColorJitter(
        brightness=0.10,
        contrast=0.10,
        saturation=0.05,
        hue=0.02
    ),
    transforms.ToTensor(),
    transforms.Normalize(
        [0.485, 0.456, 0.406],
        [0.229, 0.224, 0.225]
    )
])

# -----------------------------
# VALIDATION TRANSFORM
# -----------------------------
eval_transform = transforms.Compose([
    transforms.Lambda(lambda img: img.convert("RGB")),
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(
        [0.485, 0.456, 0.406],
        [0.229, 0.224, 0.225]
    )
])

# -----------------------------
# DATASETS
# -----------------------------
train_dataset_resize = datasets.ImageFolder(
    os.path.join(split_dir, "train"),
    transform=train_transform_resize
)

val_dataset = datasets.ImageFolder(
    os.path.join(split_dir, "val"),
    transform=eval_transform
)

# -----------------------------
# DATALOADERS
# -----------------------------
train_loader_resize = DataLoader(
    train_dataset_resize,
    batch_size=32,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

# -----------------------------
# MODEL
# -----------------------------
from torchvision import models

model = models.efficientnet_b0(weights="DEFAULT")
model.classifier[1] = nn.Linear(
    model.classifier[1].in_features,
    2
)

# -----------------------------
# LOAD ORIGINAL BASELINE
# -----------------------------
checkpoint = torch.load(
    checkpoint_path,
    map_location=device,
    weights_only=False
)

model.load_state_dict(checkpoint["model_state_dict"])
model = model.to(device)

for param in model.parameters():
    param.requires_grad = True

# -----------------------------
# CLASS WEIGHTS
# -----------------------------
class_weights = torch.tensor(
    [0.8367, 1.2425],
    dtype=torch.float32,
    device=device
)

criterion = nn.CrossEntropyLoss(
    weight=class_weights
)

# -----------------------------
# OPTIMIZER + SCHEDULER
# -----------------------------
optimizer = optim.AdamW(
    model.parameters(),
    lr=1e-4,
    weight_decay=1e-4
)

scheduler = optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="max",
    factor=0.5,
    patience=2
)

# -----------------------------
# TRAINING SETTINGS
# -----------------------------
num_epochs = 15
early_stop_patience = 4

best_val_auc = 0.0
epochs_without_improvement = 0

# -----------------------------
# TRAIN
# -----------------------------
print("Starting resize experiment...")
print("Baseline validation AUC:", checkpoint["best_val_auc"])
print("Device:", device)
print()

for epoch in range(num_epochs):

    # ===== TRAIN =====
    model.train()

    train_loss_total = 0.0
    train_correct = 0
    train_total = 0

    for images, labels in train_loader_resize:

        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(images)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        train_loss_total += loss.item() * images.size(0)

        predictions = outputs.argmax(dim=1)

        train_correct += (
            predictions == labels
        ).sum().item()

        train_total += labels.size(0)

    train_loss = train_loss_total / train_total
    train_acc = train_correct / train_total

    # ===== VALIDATION =====
    model.eval()

    val_loss_total = 0.0
    val_correct = 0
    val_total = 0

    val_probs = []
    val_labels = []

    with torch.no_grad():

        for images, labels in val_loader:

            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)

            loss = criterion(outputs, labels)

            val_loss_total += (
                loss.item() * images.size(0)
            )

            predictions = outputs.argmax(dim=1)

            val_correct += (
                predictions == labels
            ).sum().item()

            val_total += labels.size(0)

            # Class 0 = Anemic
            probs = torch.softmax(
                outputs, dim=1
            )[:, 0]

            val_probs.extend(
                probs.cpu().numpy()
            )

            val_labels.extend(
                labels.cpu().numpy()
            )

    val_loss = val_loss_total / val_total
    val_acc = val_correct / val_total

    val_auc = roc_auc_score(
        np.array(val_labels) == 0,
        np.array(val_probs)
    )

    scheduler.step(val_auc)

    lr = optimizer.param_groups[0]["lr"]

    print(
        f"Epoch {epoch+1:02d}/{num_epochs} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_acc:.4f} | "
        f"Val AUC: {val_auc:.4f} | "
        f"LR: {lr:.2e}"
    )

    # ===== SAVE BEST =====
    if val_auc > best_val_auc:

        best_val_auc = val_auc
        epochs_without_improvement = 0

        torch.save(
            {
                "epoch": epoch + 1,
                "model_state_dict": model.state_dict(),
                "optimizer_state_dict": optimizer.state_dict(),
                "val_auc": val_auc,
                "val_acc": val_acc
            },
            resize_checkpoint_path
        )

        print(
            f"  ✓ Best model saved: "
            f"{val_auc:.4f}"
        )

    else:
        epochs_without_improvement += 1

    # ===== EARLY STOPPING =====
    if epochs_without_improvement >= early_stop_patience:

        print("Early stopping triggered.")
        break

# -----------------------------
# FINAL RESULT
# -----------------------------
print()
print("=" * 55)
print("RESIZE EXPERIMENT COMPLETE")
print("=" * 55)
print(f"Baseline Val AUC : {checkpoint['best_val_auc']:.4f}")
print(f"Resize Val AUC   : {best_val_auc:.4f}")
print(f"Saved to         : {resize_checkpoint_path}")
```

```python
# ============================================================
# SWASTHYAAI — FINAL CP-ANEMIC BASELINE EVALUATION
# ============================================================

# This cell is self-contained.
# It:
# 1. Mounts Drive
# 2. Loads the existing CP-AnemiC split
# 3. Loads the original best EfficientNet-B0 checkpoint
# 4. Evaluates the LOCKED test set
# 5. Calculates final metrics
# 6. Generates confusion matrix + ROC curve
# 7. Saves final metrics
# 8. Provides a photo-only prediction function
#
# IMPORTANT:
# Test set is evaluated ONCE here.
# Do NOT use the test results to tune the model afterward.

# ============================================================
# 1. IMPORTS
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

from google.colab import drive
from torchvision import transforms, datasets, models
from torch.utils.data import DataLoader

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    roc_auc_score,
    roc_curve,
    precision_score,
    recall_score,
    f1_score
)

# ============================================================
# 2. MOUNT DRIVE
# ============================================================

drive.mount("/content/drive", force_remount=False)

# ============================================================
# 3. PATHS
# ============================================================

split_dir = "/content/CP-AnemiC_split"

checkpoint_path = (
    "/content/drive/MyDrive/Epics/AyurAI/datasets/"
    "SwasthyaAI_checkpoints/best_model_checkpoint.pth"
)

results_dir = (
    "/content/drive/MyDrive/Epics/AyurAI/datasets/"
    "SwasthyaAI_checkpoints/final_results"
)

os.makedirs(results_dir, exist_ok=True)

print("Split directory:", split_dir)
print("Checkpoint:", checkpoint_path)
print("Results directory:", results_dir)

# ============================================================
# 4. VERIFY DATASET
# ============================================================

if not os.path.exists(split_dir):
    raise FileNotFoundError(
        "CP-AnemiC split not found at /content/CP-AnemiC_split"
    )

if not os.path.exists(checkpoint_path):
    raise FileNotFoundError(
        "Original baseline checkpoint not found."
    )

# ============================================================
# 5. DEVICE
# ============================================================

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("\nDevice:", device)

# ============================================================
# 6. TEST TRANSFORM
# ============================================================

eval_transform = transforms.Compose([
    transforms.Lambda(lambda img: img.convert("RGB")),
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(
        [0.485, 0.456, 0.406],
        [0.229, 0.224, 0.225]
    )
])

# ============================================================
# 7. LOAD TEST DATASET
# ============================================================

test_dataset = datasets.ImageFolder(
    os.path.join(split_dir, "test"),
    transform=eval_transform
)

test_loader = DataLoader(
    test_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

print("\nTest samples:", len(test_dataset))
print("Classes:", test_dataset.class_to_idx)

# ============================================================
# 8. LOAD EFFICIENTNET-B0
# ============================================================

model = models.efficientnet_b0(weights="DEFAULT")

model.classifier[1] = nn.Linear(
    model.classifier[1].in_features,
    2
)

# ============================================================
# 9. LOAD ORIGINAL BEST CHECKPOINT
# ============================================================

checkpoint = torch.load(
    checkpoint_path,
    map_location=device,
    weights_only=False
)

model.load_state_dict(
    checkpoint["model_state_dict"]
)

model = model.to(device)
model.eval()

print("\nOriginal baseline checkpoint loaded.")
print(
    "Saved best validation AUC:",
    checkpoint["best_val_auc"]
)

# ============================================================
# 10. FINAL TEST EVALUATION
# ============================================================

all_labels = []
all_predictions = []
all_probabilities = []

with torch.no_grad():

    for images, labels in test_loader:

        images = images.to(device)

        outputs = model(images)

        probabilities = torch.softmax(
            outputs,
            dim=1
        )

        # Class 0 = Anemic
        anemia_probability = probabilities[:, 0]

        predictions = outputs.argmax(dim=1)

        all_labels.extend(
            labels.numpy()
        )

        all_predictions.extend(
            predictions.cpu().numpy()
        )

        all_probabilities.extend(
            anemia_probability.cpu().numpy()
        )

y_true = np.array(all_labels)
y_pred = np.array(all_predictions)
y_prob = np.array(all_probabilities)

# ============================================================
# 11. METRICS
# ============================================================

accuracy = accuracy_score(
    y_true,
    y_pred
)

# Positive class = Anemic (0)
cm = confusion_matrix(
    y_true,
    y_pred,
    labels=[1, 0]
)

# Using [Non-anemic, Anemic] ordering:
# [[TN, FP],
#  [FN, TP]]

tn, fp, fn, tp = cm.ravel()

sensitivity = tp / (tp + fn) if (tp + fn) > 0 else 0
specificity = tn / (tn + fp) if (tn + fp) > 0 else 0

precision = precision_score(
    y_true,
    y_pred,
    pos_label=0,
    zero_division=0
)

f1 = f1_score(
    y_true,
    y_pred,
    pos_label=0,
    zero_division=0
)

roc_auc = roc_auc_score(
    y_true == 0,
    y_prob
)

# ============================================================
# 12. PRINT FINAL METRICS
# ============================================================

print("\n")
print("=" * 60)
print("SWASTHYAAI — FINAL CP-ANEMIC TEST RESULTS")
print("=" * 60)

print(f"Accuracy:     {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"Sensitivity:  {sensitivity:.4f} ({sensitivity*100:.2f}%)")
print(f"Specificity:  {specificity:.4f} ({specificity*100:.2f}%)")
print(f"Precision:    {precision:.4f} ({precision*100:.2f}%)")
print(f"F1 Score:     {f1:.4f}")
print(f"ROC-AUC:      {roc_auc:.4f}")

print("\nConfusion Matrix")
print("(Rows: Actual, Columns: Predicted)")
print("Order: [Non-anemic, Anemic]")
print(cm)

# ============================================================
# 13. CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report:\n")

report = classification_report(
    y_true,
    y_pred,
    labels=[1, 0],
    target_names=[
        "Non-anemic",
        "Anemic"
    ],
    zero_division=0
)

print(report)

# ============================================================
# 14. CONFUSION MATRIX PLOT
# ============================================================

plt.figure(figsize=(6, 5))

plt.imshow(cm)

plt.title("CP-AnemiC Test Set — Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")

plt.xticks(
    [0, 1],
    ["Non-anemic", "Anemic"]
)

plt.yticks(
    [0, 1],
    ["Non-anemic", "Anemic"]
)

for i in range(2):
    for j in range(2):
        plt.text(
            j,
            i,
            cm[i, j],
            ha="center",
            va="center",
            fontsize=14
        )

plt.colorbar()
plt.tight_layout()

cm_path = os.path.join(
    results_dir,
    "final_confusion_matrix.png"
)

plt.savefig(
    cm_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 15. ROC CURVE
# ============================================================

fpr, tpr, thresholds = roc_curve(
    y_true == 0,
    y_prob
)

plt.figure(figsize=(6, 5))

plt.plot(
    fpr,
    tpr,
    label=f"EfficientNet-B0 (AUC = {roc_auc:.3f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("CP-AnemiC Test Set — ROC Curve")

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

roc_path = os.path.join(
    results_dir,
    "final_roc_curve.png"
)

plt.savefig(
    roc_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 16. SAVE METRICS
# ============================================================

final_metrics = {
    "model": "EfficientNet-B0",
    "dataset": "CP-AnemiC",
    "test_samples": int(len(y_true)),
    "accuracy": float(accuracy),
    "sensitivity": float(sensitivity),
    "specificity": float(specificity),
    "precision": float(precision),
    "f1_score": float(f1),
    "roc_auc": float(roc_auc),
    "confusion_matrix": cm.tolist(),
    "validation_auc": float(
        checkpoint["best_val_auc"]
    )
}

metrics_path = os.path.join(
    results_dir,
    "final_metrics.json"
)

with open(
    metrics_path,
    "w"
) as f:
    json.dump(
        final_metrics,
        f,
        indent=4
    )

# ============================================================
# 17. SAVE MODEL COPY
# ============================================================

final_model_path = os.path.join(
    results_dir,
    "SwasthyaAI_EfficientNetB0_CP_AnemiC.pth"
)

torch.save(
    {
        "model_state_dict": model.state_dict(),
        "model_name": "EfficientNet-B0",
        "dataset": "CP-AnemiC",
        "validation_auc": float(
            checkpoint["best_val_auc"]
        ),
        "test_accuracy": float(accuracy),
        "test_auc": float(roc_auc)
    },
    final_model_path
)

# ============================================================
# 18. PHOTO-ONLY PREDICTION FUNCTION
# ============================================================

def predict_anemia(image_path):

    from PIL import Image

    image = Image.open(
        image_path
    ).convert("RGB")

    image_tensor = eval_transform(
        image
    ).unsqueeze(0).to(device)

    model.eval()

    with torch.no_grad():

        output = model(
            image_tensor
        )

        probabilities = torch.softmax(
            output,
            dim=1
        )[0]

    anemia_probability = (
        probabilities[0].item()
    )

    non_anemia_probability = (
        probabilities[1].item()
    )

    if anemia_probability >= 0.50:
        prediction = "Anemia risk"
    else:
        prediction = "Non-anemia risk"

    return {
        "prediction": prediction,
        "anemia_probability": anemia_probability,
        "non_anemia_probability": non_anemia_probability
    }

# ============================================================
# 19. FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 60)
print("FINAL PIPELINE READY")
print("=" * 60)

print(f"Validation AUC : {checkpoint['best_val_auc']:.4f}")
print(f"Test Accuracy  : {accuracy:.4f}")
print(f"Test ROC-AUC   : {roc_auc:.4f}")

print("\nSaved files:")
print("•", cm_path)
print("•", roc_path)
print("•", metrics_path)
print("•", final_model_path)

print("\nPhoto-only prediction:")
print("result = predict_anemia('/path/to/student_photo.jpg')")
print("print(result)")

print("\n⚠️ This model is a screening/risk model, NOT a clinical diagnosis.")
```

```python
# ============================================================
# SWASTHYAAI — FINAL GRAD-CAM
# ============================================================

import os
import numpy as np
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

from PIL import Image
from torchvision import transforms, models

from pytorch_grad_cam import GradCAM
from pytorch_grad_cam.utils.model_targets import ClassifierOutputTarget
from pytorch_grad_cam.utils.image import show_cam_on_image

# ------------------------------------------------------------
# 1. DEVICE
# ------------------------------------------------------------

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)

# ------------------------------------------------------------
# 2. PATHS
# ------------------------------------------------------------

checkpoint_path = (
    "/content/drive/MyDrive/Epics/AyurAI/datasets/"
    "SwasthyaAI_checkpoints/best_model_checkpoint.pth"
)

test_root = "/content/CP-AnemiC_split/test"

results_dir = (
    "/content/drive/MyDrive/Epics/AyurAI/datasets/"
    "SwasthyaAI_checkpoints/final_results"
)

os.makedirs(results_dir, exist_ok=True)

# ------------------------------------------------------------
# 3. LOAD MODEL
# ------------------------------------------------------------

model = models.efficientnet_b0(weights="DEFAULT")

model.classifier[1] = nn.Linear(
    model.classifier[1].in_features,
    2
)

checkpoint = torch.load(
    checkpoint_path,
    map_location=device,
    weights_only=False
)

model.load_state_dict(
    checkpoint["model_state_dict"]
)

model = model.to(device)
model.eval()

print(
    "Baseline checkpoint loaded."
)
print(
    "Validation AUC:",
    checkpoint["best_val_auc"]
)

# ------------------------------------------------------------
# 4. FIND A TEST IMAGE
# ------------------------------------------------------------

image_path = None
actual_class = None

for class_name in ["Anemic", "Non-anemic"]:

    class_dir = os.path.join(
        test_root,
        class_name
    )

    files = sorted([
        f for f in os.listdir(class_dir)
        if f.lower().endswith(
            (".png", ".jpg", ".jpeg")
        )
    ])

    if files:
        image_path = os.path.join(
            class_dir,
            files[0]
        )

        actual_class = class_name
        break

if image_path is None:
    raise FileNotFoundError(
        "No test image found."
    )

print("\nImage:", image_path)
print("Actual class:", actual_class)

# ------------------------------------------------------------
# 5. LOAD IMAGE
# ------------------------------------------------------------

image = Image.open(
    image_path
).convert("RGB")

# ------------------------------------------------------------
# 6. PREPROCESS
# ------------------------------------------------------------

transform = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(
        [0.485, 0.456, 0.406],
        [0.229, 0.224, 0.225]
    )
])

input_tensor = transform(
    image
).unsqueeze(0).to(device)

# ------------------------------------------------------------
# 7. PREDICTION
# ------------------------------------------------------------

with torch.no_grad():

    output = model(
        input_tensor
    )

    probabilities = torch.softmax(
        output,
        dim=1
    )[0]

predicted_class = output.argmax(
    dim=1
).item()

anemia_probability = probabilities[0].item()
non_anemia_probability = probabilities[1].item()

class_names = {
    0: "Anemic",
    1: "Non-anemic"
}

print("\nPrediction:", class_names[predicted_class])
print(
    f"Anemia probability: "
    f"{anemia_probability:.4f}"
)
print(
    f"Non-anemia probability: "
    f"{non_anemia_probability:.4f}"
)

# ------------------------------------------------------------
# 8. GRAD-CAM
# ------------------------------------------------------------

target_layers = [
    model.features[-1]
]

cam = GradCAM(
    model=model,
    target_layers=target_layers
)

targets = [
    ClassifierOutputTarget(
        predicted_class
    )
]

grayscale_cam = cam(
    input_tensor=input_tensor,
    targets=targets
)[0]

# ------------------------------------------------------------
# 9. PREPARE IMAGE
# ------------------------------------------------------------

display_image = image.resize(
    (224, 224)
)

rgb_image = (
    np.array(display_image)
    .astype(np.float32)
    / 255.0
)

# ------------------------------------------------------------
# 10. CREATE GRAD-CAM OVERLAY
# ------------------------------------------------------------

visualization = show_cam_on_image(
    rgb_image,
    grayscale_cam,
    use_rgb=True
)

# ------------------------------------------------------------
# 11. DISPLAY
# ------------------------------------------------------------

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

plt.imshow(rgb_image)

plt.title(
    f"Original\n"
    f"Actual: {actual_class}\n"
    f"Predicted: {class_names[predicted_class]}"
)

plt.axis("off")

plt.subplot(1, 2, 2)

plt.imshow(visualization)

plt.title(
    "Grad-CAM — Model Attention"
)

plt.axis("off")

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 12. SAVE
# ------------------------------------------------------------

save_path = os.path.join(
    results_dir,
    "gradcam_example.png"
)

plt.figure(figsize=(6, 6))

plt.imshow(visualization)

plt.axis("off")

plt.title(
    "SwasthyaAI — Grad-CAM"
)

plt.tight_layout()

plt.savefig(
    save_path,
    dpi=300,
    bbox_inches="tight"
)

plt.close()

print("\nGrad-CAM saved:")
print(save_path)
```

```python
# ============================================================
# 530 DATASET — LABEL + DUPLICATE/SOURCE INSPECTION
# ============================================================

import os
import re
from collections import Counter

root = "/content/SwasthyaAI_530"

splits = ["train", "valid", "test"]

for split in splits:

    folder = os.path.join(root, split)

    files = [
        f for f in os.listdir(folder)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    anemia = [
        f for f in files
        if f.lower().startswith("anemia")
    ]

    non_anemia = [
        f for f in files
        if f.lower().startswith("non-anemia")
    ]

    print("\n" + "=" * 60)
    print(split.upper())
    print("=" * 60)

    print("Total images :", len(files))
    print("Anemia       :", len(anemia))
    print("Non-anemia   :", len(non_anemia))

    # Extract original source ID.
    # Example:
    # anemia0052_jpg.rf.abc123.jpg
    #              ↓
    # anemia0052
    #
    # Non-anemia0007_jpg.rf.xyz.jpg
    #              ↓
    # Non-anemia0007

    source_ids = []

    for f in files:

        match = re.match(
            r"(.+?)_jpg\.rf\.",
            f,
            re.IGNORECASE
        )

        if match:
            source_ids.append(
                match.group(1).lower()
            )

    counts = Counter(source_ids)

    duplicates = {
        k: v
        for k, v in counts.items()
        if v > 1
    }

    print(
        "Unique source IDs:",
        len(counts)
    )

    print(
        "Source IDs with >1 image:",
        len(duplicates)
    )

    if duplicates:

        print("\nExample duplicated source IDs:")

        for k, v in list(
            duplicates.items()
        )[:10]:

            print(
                f"  {k}: {v} images"
            )

print("\n" + "=" * 60)
print("CROSS-SPLIT SOURCE CHECK")
print("=" * 60)

split_sources = {}

for split in splits:

    folder = os.path.join(root, split)

    files = [
        f for f in os.listdir(folder)
        if f.lower().endswith(
            (".jpg", ".jpeg", ".png")
        )
    ]

    ids = set()

    for f in files:

        match = re.match(
            r"(.+?)_jpg\.rf\.",
            f,
            re.IGNORECASE
        )

        if match:
            ids.add(
                match.group(1).lower()
            )

    split_sources[split] = ids

for a, b in [
    ("train", "valid"),
    ("train", "test"),
    ("valid", "test")
]:

    overlap = (
        split_sources[a]
        &
        split_sources[b]
    )

    print(
        f"{a} ↔ {b}: "
        f"{len(overlap)} overlapping source IDs"
    )

    if overlap:

        print(
            "⚠️ Example:",
            list(overlap)[:10]
        )

print("\nInspection complete.")
```

```python
# ============================================================
# SWASTHYAAI — 530 DATASET TRAINING
# Filename-based Anemia / Non-anemia labels
# EfficientNet-B0
# ============================================================

import os
import random
import numpy as np
import torch
import torch.nn as nn

from PIL import Image
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# ============================================================
# 1. SETTINGS
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)

ROOT = "/content/SwasthyaAI_530"

# ============================================================
# 2. DATASET
# ============================================================

class AnemiaFilenameDataset(Dataset):

    def __init__(self, folder, transform=None):

        self.folder = folder
        self.transform = transform

        self.files = [
            f for f in os.listdir(folder)
            if f.lower().endswith(
                (".jpg", ".jpeg", ".png")
            )
        ]

        self.samples = []

        for filename in self.files:

            lower = filename.lower()

            # IMPORTANT:
            # Check Non-anemia FIRST because
            # "non-anemia" contains "anemia".

            if lower.startswith("non-anemia"):
                label = 1

            elif lower.startswith("anemia"):
                label = 0

            else:
                continue

            self.samples.append(
                (filename, label)
            )

        self.class_to_idx = {
            "Anemia": 0,
            "Non-anemia": 1
        }

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, index):

        filename, label = self.samples[index]

        path = os.path.join(
            self.folder,
            filename
        )

        image = Image.open(
            path
        ).convert("RGB")

        if self.transform:
            image = self.transform(image)

        return image, label


# ============================================================
# 3. TRANSFORMS
# ============================================================

train_transform = transforms.Compose([

    transforms.RandomResizedCrop(
        224,
        scale=(0.85, 1.0)
    ),

    transforms.RandomHorizontalFlip(
        p=0.5
    ),

    transforms.RandomRotation(
        5
    ),

    transforms.ColorJitter(
        brightness=0.10,
        contrast=0.10,
        saturation=0.05,
        hue=0.02
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        [0.485, 0.456, 0.406],
        [0.229, 0.224, 0.225]
    )
])

eval_transform = transforms.Compose([

    transforms.Resize(256),

    transforms.CenterCrop(224),

    transforms.ToTensor(),

    transforms.Normalize(
        [0.485, 0.456, 0.406],
        [0.229, 0.224, 0.225]
    )
])

# ============================================================
# 4. LOAD DATA
# ============================================================

train_dataset = AnemiaFilenameDataset(
    os.path.join(ROOT, "train"),
    transform=train_transform
)

val_dataset = AnemiaFilenameDataset(
    os.path.join(ROOT, "valid"),
    transform=eval_transform
)

test_dataset = AnemiaFilenameDataset(
    os.path.join(ROOT, "test"),
    transform=eval_transform
)

print("\nClass mapping:")
print(train_dataset.class_to_idx)

print("\nDataset sizes:")
print("Train:", len(train_dataset))
print("Valid:", len(val_dataset))
print("Test :", len(test_dataset))

# ============================================================
# 5. CLASS COUNTS
# ============================================================

train_labels = [
    label
    for _, label in train_dataset.samples
]

counts = np.bincount(
    train_labels,
    minlength=2
)

print("\nTraining class counts:")
print("Anemia:", counts[0])
print("Non-anemia:", counts[1])

# Class weights
class_weights = (
    len(train_labels)
    /
    (
        2 * counts
    )
)

class_weights = torch.tensor(
    class_weights,
    dtype=torch.float32
).to(device)

print("\nClass weights:")
print(class_weights)

# ============================================================
# 6. DATALOADERS
# ============================================================

train_loader = DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

# ============================================================
# 7. MODEL
# ============================================================

model = models.efficientnet_b0(
    weights="DEFAULT"
)

model.classifier[1] = nn.Linear(
    model.classifier[1].in_features,
    2
)

model = model.to(device)

# ============================================================
# 8. LOSS + OPTIMIZER
# ============================================================

criterion = nn.CrossEntropyLoss(
    weight=class_weights
)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=1e-4,
    weight_decay=1e-4
)

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="max",
    factor=0.5,
    patience=2
)

# ============================================================
# 9. VALIDATION
# ============================================================

def evaluate(loader):

    model.eval()

    labels_all = []
    anemia_probs = []

    total_loss = 0

    with torch.no_grad():

        for images, labels in loader:

            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)

            loss = criterion(
                outputs,
                labels
            )

            total_loss += (
                loss.item()
                * images.size(0)
            )

            probabilities = torch.softmax(
                outputs,
                dim=1
            )

            # Class 0 = Anemia
            probs = probabilities[:, 0]

            labels_all.extend(
                labels.cpu().numpy()
            )

            anemia_probs.extend(
                probs.cpu().numpy()
            )

    avg_loss = (
        total_loss /
        len(loader.dataset)
    )

    auc = roc_auc_score(
        labels_all,
        anemia_probs
    )

    return avg_loss, auc


# ============================================================
# 10. TRAIN
# ============================================================

EPOCHS = 20

best_val_auc = -1

best_model_path = (
    "/content/SwasthyaAI_530_best.pth"
)

print("\n" + "=" * 60)
print("STARTING TRAINING")
print("=" * 60)

for epoch in range(EPOCHS):

    model.train()

    running_loss = 0
    correct = 0
    total = 0

    for images, labels in train_loader:

        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(images)

        loss = criterion(
            outputs,
            labels
        )

        loss.backward()

        optimizer.step()

        running_loss += (
            loss.item()
            * images.size(0)
        )

        predictions = outputs.argmax(
            dim=1
        )

        correct += (
            predictions == labels
        ).sum().item()

        total += labels.size(0)

    train_loss = (
        running_loss /
        len(train_loader.dataset)
    )

    train_acc = (
        correct / total
    )

    val_loss, val_auc = evaluate(
        val_loader
    )

    scheduler.step(
        val_auc
    )

    print(
        f"Epoch {epoch+1:02d}/{EPOCHS} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val AUC: {val_auc:.4f}"
    )

    if val_auc > best_val_auc:

        best_val_auc = val_auc

        torch.save(
            {
                "model_state_dict":
                    model.state_dict(),

                "class_to_idx":
                    train_dataset.class_to_idx,

                "val_auc":
                    val_auc
            },
            best_model_path
        )

        print(
            f"  ✓ BEST MODEL SAVED "
            f"(Val AUC = {val_auc:.4f})"
        )

# ============================================================
# 11. LOAD BEST MODEL
# ============================================================

checkpoint = torch.load(
    best_model_path,
    map_location=device,
    weights_only=False
)

model.load_state_dict(
    checkpoint["model_state_dict"]
)

model.eval()

print(
    "\nBest validation AUC:",
    f"{checkpoint['val_auc']:.4f}"
)

# ============================================================
# 12. FINAL TEST EVALUATION
# ============================================================

all_labels = []
all_preds = []
all_probs = []

with torch.no_grad():

    for images, labels in test_loader:

        images = images.to(device)

        outputs = model(images)

        probabilities = torch.softmax(
            outputs,
            dim=1
        )

        predictions = outputs.argmax(
            dim=1
        )

        all_labels.extend(
            labels.numpy()
        )

        all_preds.extend(
            predictions.cpu().numpy()
        )

        all_probs.extend(
            probabilities[:, 0]
            .cpu()
            .numpy()
        )

# ============================================================
# 13. METRICS
# ============================================================

accuracy = accuracy_score(
    all_labels,
    all_preds
)

precision = precision_score(
    all_labels,
    all_preds,
    pos_label=0,
    zero_division=0
)

sensitivity = recall_score(
    all_labels,
    all_preds,
    pos_label=0,
    zero_division=0
)

f1 = f1_score(
    all_labels,
    all_preds,
    pos_label=0,
    zero_division=0
)

auc = roc_auc_score(
    all_labels,
    all_probs
)

cm = confusion_matrix(
    all_labels,
    all_preds,
    labels=[0, 1]
)

print("\n" + "=" * 60)
print("FINAL TEST RESULTS")
print("=" * 60)

print(
    f"Accuracy:     {accuracy:.4f} "
    f"({accuracy*100:.2f}%)"
)

print(
    f"Sensitivity:  {sensitivity:.4f} "
    f"({sensitivity*100:.2f}%)"
)

print(
    f"Precision:    {precision:.4f} "
    f"({precision*100:.2f}%)"
)

print(
    f"F1 Score:     {f1:.4f}"
)

print(
    f"ROC-AUC:      {auc:.4f}"
)

print("\nConfusion Matrix")
print(
    "Rows = Actual [Anemia, Non-anemia]"
)
print(
    "Cols = Predicted [Anemia, Non-anemia]"
)
print(cm)

print("\nClassification Report:")

print(
    classification_report(
        all_labels,
        all_preds,
        labels=[0, 1],
        target_names=[
            "Anemia",
            "Non-anemia"
        ],
        zero_division=0
    )
)

# ============================================================
# 14. SAVE FINAL MODEL
# ============================================================

final_model_path = (
    "/content/SwasthyaAI_530_EfficientNetB0.pth"
)

torch.save(
    {
        "model_state_dict":
            model.state_dict(),

        "class_to_idx":
            train_dataset.class_to_idx,

        "test_accuracy":
            accuracy,

        "test_sensitivity":
            sensitivity,

        "test_precision":
            precision,

        "test_f1":
            f1,

        "test_auc":
            auc
    },
    final_model_path
)

print(
    "\n✓ FINAL MODEL SAVED:"
)

print(final_model_path)

print("\nDONE.")
```

```python
# ============================================================
# SWASTHYAAI — 530 IMAGE ANEMIA CLASSIFICATION
# EfficientNet-B0 | Correct Anemia AUC
# ============================================================

import os
import random
import numpy as np
import pandas as pd
from PIL import Image

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models

from sklearn.metrics import roc_auc_score

# ============================================================
# 1. REPRODUCIBILITY
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False


# ============================================================
# 2. DEVICE
# ============================================================

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Device:", device)


# ============================================================
# 3. DATASET PATHS
# ============================================================

DATA_ROOT = "/content/SwasthyaAI_530"

TRAIN_DIR = os.path.join(DATA_ROOT, "train")
VAL_DIR   = os.path.join(DATA_ROOT, "valid")
TEST_DIR  = os.path.join(DATA_ROOT, "test")


# ============================================================
# 4. CLASS MAPPING
# ============================================================
# IMPORTANT:
# Anemia = class 0
# Non-anemia = class 1
#
# For ROC-AUC we will use:
# probability of Anemia = probability of class 0
# ============================================================

class_to_idx = {
    "Anemia": 0,
    "Non-anemia": 1
}

idx_to_class = {
    0: "Anemia",
    1: "Non-anemia"
}

print("\nClass mapping:")
print(class_to_idx)


# ============================================================
# 5. CUSTOM DATASET
# ============================================================

class AnemiaFilenameDataset(Dataset):

    def __init__(self, folder, transform=None):

        self.folder = folder
        self.transform = transform

        self.samples = []

        for filename in sorted(os.listdir(folder)):

            filepath = os.path.join(folder, filename)

            # Ignore non-image files
            if not filename.lower().endswith(
                (".jpg", ".jpeg", ".png", ".bmp", ".webp")
            ):
                continue

            # IMPORTANT:
            # Check Non-anemia FIRST because
            # "Non-anemia" also contains the word "anemia".

            if filename.lower().startswith("non-anemia"):
                label = class_to_idx["Non-anemia"]

            elif filename.lower().startswith("anemia"):
                label = class_to_idx["Anemia"]

            else:
                continue

            self.samples.append((filepath, label))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, index):

        filepath, label = self.samples[index]

        image = Image.open(filepath).convert("RGB")

        if self.transform:
            image = self.transform(image)

        return image, torch.tensor(label, dtype=torch.long)


# ============================================================
# 6. IMAGE TRANSFORMS
# ============================================================

train_transform = transforms.Compose([

    transforms.RandomResizedCrop(
        224,
        scale=(0.85, 1.0)
    ),

    transforms.RandomHorizontalFlip(
        p=0.5
    ),

    transforms.RandomRotation(
        5
    ),

    transforms.ColorJitter(
        brightness=0.10,
        contrast=0.10,
        saturation=0.05,
        hue=0.02
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


eval_transform = transforms.Compose([

    transforms.Resize(256),

    transforms.CenterCrop(224),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


# ============================================================
# 7. CREATE DATASETS
# ============================================================

train_dataset = AnemiaFilenameDataset(
    TRAIN_DIR,
    transform=train_transform
)

val_dataset = AnemiaFilenameDataset(
    VAL_DIR,
    transform=eval_transform
)

test_dataset = AnemiaFilenameDataset(
    TEST_DIR,
    transform=eval_transform
)


# ============================================================
# 8. DATASET INFORMATION
# ============================================================

def count_classes(dataset):

    labels = [label for _, label in dataset.samples]

    anemia = labels.count(0)
    non_anemia = labels.count(1)

    return anemia, non_anemia


train_anemia, train_non_anemia = count_classes(train_dataset)
val_anemia, val_non_anemia = count_classes(val_dataset)
test_anemia, test_non_anemia = count_classes(test_dataset)

print("\nDataset sizes:")
print("Train:", len(train_dataset))
print("Valid:", len(val_dataset))
print("Test :", len(test_dataset))

print("\nTraining class counts:")
print("Anemia:", train_anemia)
print("Non-anemia:", train_non_anemia)

print("\nValidation class counts:")
print("Anemia:", val_anemia)
print("Non-anemia:", val_non_anemia)

print("\nTest class counts:")
print("Anemia:", test_anemia)
print("Non-anemia:", test_non_anemia)


# ============================================================
# 9. DATALOADERS
# ============================================================

BATCH_SIZE = 32

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)


# ============================================================
# 10. CLASS WEIGHTS
# ============================================================

total_train = train_anemia + train_non_anemia

weight_anemia = total_train / (2 * train_anemia)
weight_non_anemia = total_train / (2 * train_non_anemia)

class_weights = torch.tensor(
    [weight_anemia, weight_non_anemia],
    dtype=torch.float32
).to(device)

print("\nClass weights:")
print(class_weights)


# ============================================================
# 11. MODEL — EFFICIENTNET-B0
# ============================================================

model = models.efficientnet_b0(
    weights=models.EfficientNet_B0_Weights.DEFAULT
)

model.classifier[1] = nn.Linear(
    model.classifier[1].in_features,
    2
)

model = model.to(device)


# ============================================================
# 12. LOSS FUNCTION
# ============================================================

criterion = nn.CrossEntropyLoss(
    weight=class_weights
)


# ============================================================
# 13. OPTIMIZER
# ============================================================

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=1e-4,
    weight_decay=1e-4
)


# ============================================================
# 14. LR SCHEDULER
# ============================================================

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="max",
    factor=0.5,
    patience=2
)


# ============================================================
# 15. TRAINING SETTINGS
# ============================================================

NUM_EPOCHS = 20

BEST_VAL_AUC = -float("inf")

BEST_MODEL_PATH = "/content/SwasthyaAI_530_best_correct_auc.pth"


# ============================================================
# 16. TRAINING LOOP
# ============================================================

print("\n" + "=" * 60)
print("STARTING TRAINING")
print("=" * 60)


for epoch in range(NUM_EPOCHS):

    # --------------------------------------------------------
    # TRAIN
    # --------------------------------------------------------

    model.train()

    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in train_loader:

        images = images.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        optimizer.zero_grad()

        outputs = model(images)

        loss = criterion(outputs, labels)

        loss.backward()

        optimizer.step()

        running_loss += loss.item() * images.size(0)

        predictions = torch.argmax(outputs, dim=1)

        correct += (predictions == labels).sum().item()

        total += labels.size(0)


    train_loss = running_loss / total
    train_acc = correct / total


    # --------------------------------------------------------
    # VALIDATION
    # --------------------------------------------------------

    model.eval()

    val_running_loss = 0.0

    val_labels = []
    val_anemia_probs = []

    val_correct = 0
    val_total = 0

    with torch.no_grad():

        for images, labels in val_loader:

            images = images.to(device, non_blocking=True)
            labels = labels.to(device, non_blocking=True)

            outputs = model(images)

            loss = criterion(outputs, labels)

            val_running_loss += loss.item() * images.size(0)

            probabilities = torch.softmax(
                outputs,
                dim=1
            )

            predictions = torch.argmax(
                outputs,
                dim=1
            )

            val_correct += (
                predictions == labels
            ).sum().item()

            val_total += labels.size(0)

            # ------------------------------------------------
            # IMPORTANT:
            # Class 0 = ANEMIA
            #
            # Therefore use probability[:, 0]
            # ------------------------------------------------

            anemia_probability = probabilities[:, 0]

            val_anemia_probs.extend(
                anemia_probability.cpu().numpy().tolist()
            )

            val_labels.extend(
                labels.cpu().numpy().tolist()
            )


    val_loss = val_running_loss / val_total
    val_acc = val_correct / val_total


    # --------------------------------------------------------
    # CORRECT AUC
    # --------------------------------------------------------

    val_labels_np = np.array(val_labels)

    val_anemia_probs_np = np.array(
        val_anemia_probs
    )

    val_auc = roc_auc_score(
        val_labels_np == 0,
        val_anemia_probs_np
    )


    # --------------------------------------------------------
    # SCHEDULER
    # --------------------------------------------------------

    scheduler.step(val_auc)


    # --------------------------------------------------------
    # PRINT
    # --------------------------------------------------------

    print(
        f"Epoch {epoch+1:02d}/{NUM_EPOCHS} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_acc:.4f} | "
        f"Val AUC: {val_auc:.4f}"
    )


    # --------------------------------------------------------
    # SAVE BEST MODEL
    # --------------------------------------------------------

    if val_auc > BEST_VAL_AUC:

        BEST_VAL_AUC = val_auc

        torch.save(
            {
                "epoch": epoch + 1,
                "model_state_dict": model.state_dict(),
                "optimizer_state_dict": optimizer.state_dict(),
                "val_auc": val_auc,
                "class_to_idx": class_to_idx
            },
            BEST_MODEL_PATH
        )

        print(
            f"  ✓ BEST MODEL SAVED "
            f"(Val AUC = {val_auc:.4f})"
        )


# ============================================================
# 17. TRAINING COMPLETE
# ============================================================

print("\n" + "=" * 60)
print("TRAINING COMPLETE")
print("=" * 60)

print(
    f"Best Validation AUC: {BEST_VAL_AUC:.4f}"
)

print(
    f"Best model saved at:\n{BEST_MODEL_PATH}"
)

print("\nIMPORTANT:")
print("Test set has NOT been evaluated yet.")
print("The test set remains untouched.")
```

```python
# ============================================================
# SWASTHYAAI — FINAL TEST EVALUATION
# 530-IMAGE DATASET
# BEST VALIDATION-AUC CHECKPOINT ONLY
# ============================================================

import os
import json
import numpy as np
import torch
import torch.nn as nn

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# ------------------------------------------------------------
# Load best model
# ------------------------------------------------------------

BEST_MODEL_PATH = "/content/SwasthyaAI_530_best_correct_auc.pth"

checkpoint = torch.load(
    BEST_MODEL_PATH,
    map_location=device,
    weights_only=False
)

model.load_state_dict(
    checkpoint["model_state_dict"]
)

model = model.to(device)
model.eval()

print("Loaded best model from epoch:",
      checkpoint["epoch"])

print("Best validation AUC:",
      checkpoint["val_auc"])


# ------------------------------------------------------------
# TEST SET
# ------------------------------------------------------------

test_labels = []
test_anemia_probs = []
test_predictions = []

with torch.no_grad():

    for images, labels in test_loader:

        images = images.to(
            device,
            non_blocking=True
        )

        labels = labels.to(
            device,
            non_blocking=True
        )

        outputs = model(images)

        probabilities = torch.softmax(
            outputs,
            dim=1
        )

        predictions = torch.argmax(
            outputs,
            dim=1
        )

        # Class 0 = Anemia
        anemia_probability = probabilities[:, 0]

        test_labels.extend(
            labels.cpu().numpy().tolist()
        )

        test_anemia_probs.extend(
            anemia_probability.cpu().numpy().tolist()
        )

        test_predictions.extend(
            predictions.cpu().numpy().tolist()
        )


# ------------------------------------------------------------
# CONVERT TO NUMPY
# ------------------------------------------------------------

y_true = np.array(test_labels)

y_pred = np.array(test_predictions)

y_anemia_prob = np.array(test_anemia_probs)


# ------------------------------------------------------------
# CORRECT ROC-AUC
# ------------------------------------------------------------

test_auc = roc_auc_score(
    y_true == 0,
    y_anemia_prob
)


# ------------------------------------------------------------
# ACCURACY
# ------------------------------------------------------------

accuracy = accuracy_score(
    y_true,
    y_pred
)


# ------------------------------------------------------------
# ANEMIA METRICS
# ------------------------------------------------------------

# Anemia = class 0
# Therefore explicitly convert to binary:
# Anemia = 1
# Non-anemia = 0

y_true_anemia = (y_true == 0).astype(int)
y_pred_anemia = (y_pred == 0).astype(int)

sensitivity = recall_score(
    y_true_anemia,
    y_pred_anemia,
    zero_division=0
)

precision = precision_score(
    y_true_anemia,
    y_pred_anemia,
    zero_division=0
)

f1 = f1_score(
    y_true_anemia,
    y_pred_anemia,
    zero_division=0
)


# ------------------------------------------------------------
# SPECIFICITY
# ------------------------------------------------------------

cm = confusion_matrix(
    y_true_anemia,
    y_pred_anemia
)

tn, fp, fn, tp = cm.ravel()

specificity = tn / (tn + fp) if (tn + fp) > 0 else 0


# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("FINAL TEST RESULTS")
print("=" * 60)

print(f"Accuracy             : {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"Sensitivity (Anemia) : {sensitivity:.4f} ({sensitivity*100:.2f}%)")
print(f"Specificity          : {specificity:.4f} ({specificity*100:.2f}%)")
print(f"Precision (Anemia)   : {precision:.4f} ({precision*100:.2f}%)")
print(f"F1 Score (Anemia)    : {f1:.4f}")
print(f"ROC-AUC              : {test_auc:.4f}")


# ------------------------------------------------------------
# CONFUSION MATRIX
# ------------------------------------------------------------

print("\nConfusion Matrix")
print("----------------")
print("[[TN, FP],")
print(" [FN, TP]]")

print(cm)


# ------------------------------------------------------------
# CLASSIFICATION REPORT
# ------------------------------------------------------------

print("\nClassification Report")
print("---------------------")

print(
    classification_report(
        y_true,
        y_pred,
        target_names=[
            "Anemia",
            "Non-anemia"
        ],
        zero_division=0
    )
)


# ------------------------------------------------------------
# SAVE RESULTS
# ------------------------------------------------------------

results = {
    "dataset": "SwasthyaAI 530-image Roboflow dataset",
    "model": "EfficientNet-B0",
    "best_epoch": int(checkpoint["epoch"]),
    "best_validation_auc": float(checkpoint["val_auc"]),
    "test_size": int(len(y_true)),
    "test_anemia": int((y_true == 0).sum()),
    "test_non_anemia": int((y_true == 1).sum()),
    "accuracy": float(accuracy),
    "sensitivity_anemia": float(sensitivity),
    "specificity_non_anemia": float(specificity),
    "precision_anemia": float(precision),
    "f1_anemia": float(f1),
    "roc_auc": float(test_auc),
    "confusion_matrix": cm.tolist()
}

RESULT_PATH = "/content/SwasthyaAI_530_final_test_results.json"

with open(
    RESULT_PATH,
    "w"
) as f:

    json.dump(
        results,
        f,
        indent=4
    )

print("\nResults saved to:")
print(RESULT_PATH)

print("\n" + "=" * 60)
print("TEST EVALUATION COMPLETE")
print("=" * 60)
```
